### **1.3 Dataset & Environment Setup**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style for clean storytelling
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update({'font.sans-serif': 'DejaVu Sans', 'font.size': 11})

# Load and Preprocess Dataset
df = sns.load_dataset('penguins')

# Impute missing numerical traits with median
num_cols = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

# Drop missing categorical values
df = df.dropna(subset=['sex'])

print("✅ Dataset successfully loaded & cleaned for storytelling!")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
```

## **SECTION 2: Visual Storytelling & Visualizations**

### **2.1 Visualization 1: Population Distribution (Donut Chart)**

```python
# Figure 1: Donut Chart for Species Distribution
plt.figure(figsize=(7, 7))
species_counts = df['species'].value_counts()
colors = ['#66b3ff', '#ff9999', '#99ff99']

plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%', startangle=140,
        colors=colors, wedgeprops=dict(width=0.4, edgecolor='w', linewidth=2))

plt.title('Figure 1: Species Distribution in Palmer Archipelago', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

### **Explanation & Narrative Insights:**

Visual Choice: A clean Donut Chart was selected instead of a pie chart for better clarity in showing proportions to non-technical readers.

Narrative Explanation: The dataset consists of 43.8% Adelie, 35.7% Gentoo, and 20.4% Chinstrap penguins. Adelie forms the baseline majority, while Chinstrap is the minority group.

Key Takeaway: Establishes the sample baseline, showing that species comparisons must account for the smaller sample size of Chinstrap penguins.

### **2.2 Visualization 2: Body Mass Range Across Species (Box Plot)**

```python
# Figure 2: Categorical Boxplot for Body Mass
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='species', y='body_mass_g', hue='species', palette='Set2', legend=False)

plt.title('Figure 2: Body Mass Distribution Across Penguin Species', fontsize=13, fontweight='bold')
plt.xlabel('Penguin Species', fontsize=11, fontweight='bold')
plt.ylabel('Body Mass (Grams)', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```

## **Explanation & Narrative Insights:**

Visual Choice: A Boxplot clearly shows medians, ranges, and outliers without overwhelming non-technical readers with dense scatter points.

Narrative Explanation: Gentoo penguins are distinctly heavier than Adelie and Chinstrap species. While Adelie and Chinstrap cluster tightly between 2,700g and 4,800g, Gentoo penguins average over 5,000g (ranging from 3,900g to 6,300g).

Key Takeaway: Body mass serves as the single strongest physical differentiator for identifying Gentoo penguins in ecological field studies

## **2.3 Visualization 3: Flipper Length vs Total Body Mass (Regression Plot)**

```python
# Figure 3: Scatter Plot with Linear Trendlines
plt.figure(figsize=(9, 5))
sns.lmplot(data=df, x='flipper_length_mm', y='body_mass_g', hue='species',
           palette='Set1', height=5, aspect=1.6, scatter_kws={'alpha':0.7})

plt.title('Figure 3: Flipper Length vs Total Body Mass by Species', fontsize=13, fontweight='bold')
plt.xlabel('Flipper Length (mm)', fontsize=11, fontweight='bold')
plt.ylabel('Body Mass (g)', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```

## **Explanation & Narrative Insights:**

Visual Choice: A Scatter Plot with Trendlines visualizes the linear relationship between two physical measurements across groups.Narrative Explanation: Flipper length directly tracks total body mass across all three species. As flipper length increases, body mass increases in a predictable linear pattern ($r \approx 0.87$).Key Takeaway: Flipper length acts as an excellent direct proxy for estimating total body mass, allowing field researchers to estimate mass without heavy weighing equipment.

## **2.4 Visualization 4: Gender Dimorphism in Bill Length (Violin Plot)**

```python
# Figure 4: Split Violin Plot for Sex Comparison
plt.figure(figsize=(9, 5))
sns.violinplot(data=df, x='species', y='bill_length_mm', hue='sex', split=True, palette='Pastel1', inner='quart')

plt.title('Figure 4: Sexual Dimorphism in Bill Length (Male vs Female)', fontsize=13, fontweight='bold')
plt.xlabel('Species', fontsize=11, fontweight='bold')
plt.ylabel('Bill Length (mm)', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```

## **Explanation & Narrative Insights:**

Visual Choice: A Split Violin Plot compares male vs. female distributions side-by-side in a compact, highly visual format.

Narrative Explanation: Across every species, male penguins consistently display longer bill lengths than females. Furthermore, Chinstrap and Gentoo species exhibit noticeably longer bills overall compared to Adelie penguins.

Key Takeaway: Demonstrates strong sexual dimorphism across traits, proving that demographic studies must always segment measurements by gender to avoid biased conclusions

## **2.5 Visualization 5: Physical Attribute Correlation Heatmap**

```python
# Figure 5: Feature Correlation Heatmap
plt.figure(figsize=(8, 6))
corr = df[num_cols].corr()

sns.heatmap(corr, annot=True, cmap='Blues', fmt='.2f', linewidths=1, annot_kws={"size": 12})
plt.title('Figure 5: Physical Traits Correlation Heatmap', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

## **Explanation & Narrative Insights**

Visual Choice: A Heatmap simplifies complex numerical correlation matrices into an easy-to-read color grid.

Narrative Explanation: Darker blue colors indicate strong positive correlations, while lighter shades represent weaker relationships. Flipper length and body mass show the highest correlation (0.87).

Key Takeaway: Provides a baseline feature correlation matrix for selecting variables in future machine learning models.
